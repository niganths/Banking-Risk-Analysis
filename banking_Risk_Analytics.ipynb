```python
# Age Distribution

plt.figure(figsize=(8, 5))
sns.histplot(df["Age"], bins=15, kde=True)
plt.title("Age Distribution of Customers")
plt.xlabel("Age")
plt.ylabel("Number of Customers")
plt.show()
```

```python
plt.figure(figsize=(10, 6))
df["Occupation"].value_counts().head(10).plot(kind="bar")
plt.title("Top 10 Customer Distribution by Occupation")
plt.xlabel("Occupation")
plt.ylabel("Number of Customers")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="Estimated Income", y="Bank Loans")
plt.title("Bank Loans vs Estimated Income")
plt.xlabel("Estimated Income")
plt.ylabel("Bank Loans")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x="Risk Weighting", y="Bank Loans")
plt.title("Bank Loans by Risk Weighting")
plt.xlabel("Risk Weighting")
plt.ylabel("Bank Loans")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="Bank Deposits", y="Bank Loans")
plt.title("Bank Deposits vs Bank Loans")
plt.xlabel("Bank Deposits")
plt.ylabel("Bank Loans")
plt.show()
```

```python
top_occupations = df["Occupation"].value_counts().head(10).index

risk_occupation = pd.crosstab(
    df.loc[df["Occupation"].isin(top_occupations), "Occupation"],
    df.loc[df["Occupation"].isin(top_occupations), "Risk Weighting"]
)

risk_occupation = risk_occupation.loc[top_occupations]

plt.figure(figsize=(12, 6))
sns.heatmap(risk_occupation, annot=True, fmt="d", cmap="Blues")
plt.title("Top 10 Occupations vs Risk Weighting")
plt.xlabel("Risk Weighting")
plt.ylabel("Occupation")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=df,
    x="Amount of Credit Cards",
    y="Credit Card Balance"
)
plt.title("Credit Card Balance vs Number of Credit Cards")
plt.xlabel("Number of Credit Cards")
plt.ylabel("Credit Card Balance")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_excel("merged_banking_data.xlsx")

columns = [
    "Age",
    "Estimated Income",
    "Superannuation Savings",
    "Credit Card Balance",
    "Bank Loans",
    "Bank Deposits",
    "Checking Accounts",
    "Saving Accounts",
    "Properties Owned",
    "Risk Weighting"
]

correlation = df[columns].corr()

plt.figure(figsize=(12, 8))
sns.heatmap(correlation, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap of Banking Variables")
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df,
    x="Estimated Income",
    y="Bank Loans",
    hue="Risk Weighting",
    size="Age",
    sizes=(30, 200)
)
plt.title("Income, Bank Loans and Risk Weighting")
plt.xlabel("Estimated Income")
plt.ylabel("Bank Loans")
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df,
    x="Bank Deposits",
    y="Bank Loans",
    hue="Risk Weighting",
    size="Estimated Income",
    sizes=(30, 200)
)
plt.title("Loans, Deposits, Income and Risk Weighting")
plt.xlabel("Bank Deposits")
plt.ylabel("Bank Loans")
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df,
    x="Estimated Income",
    y="Credit Card Balance",
    hue="Bank Loans",
    size="Risk Weighting",
    sizes=(30, 200)
)
plt.title("Income, Credit Card Balance, Loans and Risk")
plt.xlabel("Estimated Income")
plt.ylabel("Credit Card Balance")
plt.show()
```

```python
risk_profile = df.groupby("Risk Weighting")[
    [
        "Estimated Income",
        "Bank Loans",
        "Bank Deposits",
        "Credit Card Balance",
        "Superannuation Savings"
    ]
].mean()

plt.figure(figsize=(12, 6))
sns.heatmap(risk_profile, annot=True, fmt=".0f", cmap="YlGnBu")
plt.title("Average Financial Profile by Risk Weighting")
plt.xlabel("Financial Variables")
plt.ylabel("Risk Weighting")
plt.show()
```
